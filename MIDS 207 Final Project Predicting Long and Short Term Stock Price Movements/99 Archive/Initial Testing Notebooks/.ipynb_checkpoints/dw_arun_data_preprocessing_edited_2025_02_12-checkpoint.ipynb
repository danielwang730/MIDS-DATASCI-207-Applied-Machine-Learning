## Feature Engineering:

Features 1 through 4:
* PDC (previous day close)
* PRL (previous day low)
* PRH (previous day high)
* Growth Rate: the percentage change in a value over a specific time period, often expressed as a percentage. Growth Rate = (New Value - Old Value)/(Old Value) * 100%

```python
# Creating a new Date_col column to rank and drop dates appropriately
data = data.reset_index().set_index("Date", drop=False)
data = data.rename(columns={"Date": "Date_col"})
data["rank"] = data.groupby("Ticker")["Date_col"].rank()
data = data.sort_values(by=["Ticker", "Date_col"])

data
```

```python
# Getting difference between close price and open price
data["Close_Open_Diff"] = data["Close"] - data["Open"]
data["Close_Open_Diff_Perc"] = (data["Close"] - data["Open"]) / data["Open"] * 100

# Getting binary classification for whether a stock ended positive (1) or negative/zero (0)
data["Close_Open_Diff_Binary"] = data["Close_Open_Diff"] > 0

# Getting ternary classification for whether a stock ended positive (2), negative (0), or not much change (1)
close_open_conditions = [
    data['Close_Open_Diff_Perc'] > 0.5,       # Close price was at least 0.5% higher than Open price
    (data['Close_Open_Diff_Perc'] >= -0.5) &  # Close price was between -0.5% and 0.5% of Open price
        (data['Close_Open_Diff_Perc'] <= 0.5),
    data['Close_Open_Diff_Perc'] < -0.5       # Close price was at least 0.5% lower than Open price
]
close_open_choices = [2, 1, 0]  # Defining choices corresponding to conditions
data['Close_Open_Diff_Ternary'] = np.select(close_open_conditions, close_open_choices)  # Applying conditions

data
```

```python
# Getting growth rates and previous day OHLCV metrics
for x in data.columns[2:-6]:
  data[f'Growth Rate_{x}'] = (data[x] - data[x].shift(1))/data[x].shift(1) * 100
data["PDO"] = data["Open"].shift(1)
data["PDH"] = data["High"].shift(1)
data["PDL"] = data["Low"].shift(1)
data["PDC"] = data["Close"].shift(1)
data["PDV"] = data["Volume"].shift(1)

# For row in data, if date is 2023-05-05, set above column values to null. Note: date is the index:
data.loc[start_date_analyzed, ['PDO', 'PDH', 'PDL', 'PDC', 'PDV', 'Growth Rate_Open', 'Growth Rate_High', 'Growth Rate_Low',
    'Growth Rate_Close']] = np.nan

data
```

Derived Features:
* Gap up/gap down: (Open - PDC) / PDC; Could be either binary (e.g., |X| > 2%) or metric
* Daily range gap up: (Open - PDH) / PDH; Must be positive; if 0 or negative, then open price has not gapped up from previous day range
* Daily range gap down: (Open - PDL) / PDL; Must be negative; if 0 or positive, then open price has not gapped down from previous day range
* Average volume (14 days)


```python
# Getting the opening gap features from PDC
data["Gap_from_PDC"] = (data["Open"] - data["PDC"]) / data["PDC"]
gap_from_PDC_threshold = 0.02  # 2%
data["Gap_Binary_Threshold"] = abs(data["Gap_from_PDC"]) > gap_from_PDC_threshold

# Getting daily range gap up features
data["Daily_Range_Gap_Up"] = (data["Open"] - data["PDH"]) / data["PDH"]
data["Daily_Range_Gap_Up"] = data["Daily_Range_Gap_Up"].clip(lower = 0)
data["DR_Gap_Up_Binary"] = data["Daily_Range_Gap_Up"] > 0

# Getting daily range gap down features
data["Daily_Range_Gap_Down"] = (data["Open"] - data["PDL"]) / data["PDL"]
data["Daily_Range_Gap_Down"] = data["Daily_Range_Gap_Down"].clip(upper = 0)
data["DR_Gap_Down_Binary"] = data["Daily_Range_Gap_Down"] < 0

# Getting 14-day average volume and previous volume features
avg_vol_window = 14
data["Avg_Volume_14D"] = data["Volume"].rolling(window=avg_vol_window).mean()
data["Avg_Vol_Prev_14D"] = data["PDV"].rolling(window=avg_vol_window).mean()

# Comparing current volume to average volume over the past XX days
data["Vol_Change"] = (data["Volume"] - data["Avg_Vol_Prev_14D"]) / data["Avg_Vol_Prev_14D"]

# Setting the first 14-day volume averages and previous averages of every ticker to np.nan
data.loc[data["rank"] < 15, ["Avg_Volume_14D", "Avg_Vol_Prev_14D", "Vol_Change"]] = np.nan

data[10:20]
```

Technical/Momentum Indicators:

* SMA (simple moving average) (14-day, 50-day, 200-day): average of the last XX closing prices
* EMA (exponential moving average) (9-day, 20-day):
(Opentoday * Smoothing1 + Days) + EMAyesterday * (1 - Smoothing1 + Days)
Smoothing: usually starts with a factor of 2
Increasing the smoothing will put more emphasis on today’s open
Initial calculations for EMA yesterday can be calculated from SMA
https://www.investopedia.com/terms/e/ema.asp
https://www.strike.money/technical-analysis/ema
* RSI (relative strength index) (14-day):
“A momentum indicator used in technical analysis. RSI measures the speed and magnitude of a security’s recent price changes to detect overbought or oversold conditions in the price of that security.”
https://www.investopedia.com/terms/r/rsi.asp
RSI = 100 - 1001 + (Previous Average Gain * 13) + Current Gain(Previous Average Loss * 13) + Current Loss
Previous average gain: average of the past 13 days where the close was greater than the open
Previous average loss: average of the past 13 days where the close was lower than the open
* ATR (average true range) (14-day):
Shows the average range of the past XX days
(1/n) * Σpast 14 days (PDC - PDL)
This is a simplified version of ATR that we can use as either a binary or metric indicator
Likely needs to be standardized (an ATR of $1 might be a lot for a $10 stock, but might be nothing for a $300 stock)
* Can also consider ATR% (ATRP): (ATR / PDC) * 100
https://www.investopedia.com/terms/a/atr.asp


```python
# SMA and EMA including this period
data["SMA_14"] = data["Close"].rolling(window=14).mean()
data["SMA_50"] = data["Close"].rolling(window=50).mean()
data["SMA_200"] = data["Close"].rolling(window=200).mean()
data["EMA_9"] = data["Close"].ewm(span=9, adjust=False).mean()
data["EMA_20"] = data["Close"].ewm(span=20, adjust=False).mean()

# SMA and EMA for the past XXX periods
data["SMA_prev_14"] = data["PDC"].rolling(window=14).mean()
data["SMA_prev_50"] = data["PDC"].rolling(window=50).mean()
data["SMA_prev_200"] = data["PDC"].rolling(window=200).mean()
data["EMA_prev_9"] = data["PDC"].ewm(span=9, adjust=False).mean()
data["EMA_prev_20"] = data["PDC"].ewm(span=20, adjust=False).mean()

# RSI Calculation
delta = data["Close"].diff()
gain = (delta.where(delta > 0, 0)).rolling(window=14).mean()
loss = (-delta.where(delta < 0, 0)).rolling(window=14).mean()
rs = gain / loss
data["RSI_14"] = 100 - (100 / (1 + rs))

# RSI for the previous XX periods using tulipy
rsi_prev_14 = ti.rsi(data["PDC"].to_numpy(), 14)
nan_array_rsi = np.full(14, np.nan)
rsi_prev_14 = np.concatenate((nan_array_rsi, rsi_prev_14))
data["RSI_prev_14"] = rsi_prev_14

# ATR Calculation
data["ATR_14"] = (data["PDH"] - data["PDL"]).rolling(window=14).mean()
data["NATR_14"] = (data["ATR_14"] / data["PDC"]) * 100

# Setting 9, 14, 20, 50, and 200 day metrics to np.nan:
data.loc[data["rank"] <= 9, ['EMA_9', 'EMA_prev_9']] = np.nan
data.loc[data["rank"] <= 14, ['SMA_14', 'SMA_prev_14', 'RSI_14', 'RSI_prev_14', 'ATR_14', 'NATR_14']] = np.nan
data.loc[data["rank"] <= 20, ['EMA_20', 'EMA_prev_20']] = np.nan
data.loc[data["rank"] <= 50, ['SMA_50', 'SMA_prev_50']] = np.nan
data.loc[data["rank"] <= 200, ['SMA_200', 'SMA_prev_200']] = np.nan

data[:18]
```

```python
# Just getting the AAPL stock
data_small = data[data['Ticker'] == 'AAPL']

# Only keeping certain columns (explanations below)
data_small_columns_keep = ['Gap_from_PDC', 'Gap_Binary_Threshold', 'DR_Gap_Up_Binary', 'DR_Gap_Down_Binary', 'Vol_Change',
    'RSI_prev_14', 'Close_Open_Diff', 'Close_Open_Diff_Perc', 'Close_Open_Diff_Binary', 'Close_Open_Diff_Ternary']
# Overall, these metrics are all either binary, based on a percent, or have a set range (e.g., RSI); these qualities make the features
# generalizable across any tickers and across any day
# - Gap_from_PDC: Describes how much the open changed from the PDC, which could indicate which direction it will likely end up
# - Gap_Binary_Threshold: If it met a threshold of a gap (in the original case: 2%), perhaps it's more likely to go a certain direction
# - DR_Gap_Up_Binary: If a stock did gap up from the previous daily range, perhaps it's more likely to end positive as well
# - DR_Gap_Down_Binary: If a stock did gap down from the previous daily range, perhaps it's more likely to end negative as well
# - Vol_Change: A greater volume change might indicate more volatility from the previous day, creating more opportunities for bigger
# gaps; one thing to try later would be to separate out into longs and shorts, and see if the volume change could be a factor into how
# large the difference is between the close and the open
# - RSI_prev_14: A stock that is overbought might drop this day, and vice versa for a stock that is oversold
# - Close_Open_Diff: Potential outcome variable
# - Close_Open_Diff_Perc: Potential outcome variable
# - Close_Open_Diff_Binary: Potential outcome variable
# - Close_Open_Diff_Ternary: Potential outcome variable
# - 
# - Comment about NATR_14: Included initially, but ultimately kept it out because it just determines entry conditions, not overall
# performance

data_small = data_small[data_small_columns_keep]

data_small
```

```python
# Standardizing Gap_from_PDC and Vol_Change while normalizing RSI_prev_14

# Getting the mean and std of Gap_from_PDC and Vol_Change
X_train_gap_mean = X_train['Gap_from_PDC'].mean()
X_train_gap_std = X_train['Gap_from_PDC'].std()
X_train_vol_mean = X_train['Vol_Change'].mean()
X_train_vol_std = X_train['Vol_Change'].std()

# Standardizing Gap_from_PDC and Vol_Change for all three datasets
X_train["Gap_from_PDC"] = (X_train["Gap_from_PDC"] - X_train_gap_mean) / X_train_gap_std
X_train["Vol_Change"] = (X_train["Vol_Change"] - X_train_vol_mean) / X_train_vol_std

X_val["Gap_from_PDC"] = (X_val["Gap_from_PDC"] - X_train_gap_mean) / X_train_gap_std
X_val["Vol_Change"] = (X_val["Vol_Change"] - X_train_vol_mean) / X_train_vol_std

X_test["Gap_from_PDC"] = (X_test["Gap_from_PDC"] - X_train_gap_mean) / X_train_gap_std
X_test["Vol_Change"] = (X_test["Vol_Change"] - X_train_vol_mean) / X_train_vol_std

# Normalizing RSI_prev_14 to 100 (RSI can only range from 0 to 100)
X_train["RSI_prev_14"] = X_train["RSI_prev_14"] / 100
X_val["RSI_prev_14"] = X_val["RSI_prev_14"] / 100
X_test["RSI_prev_14"] = X_test["RSI_prev_14"] / 100

# Checking to see if X_train makes sense and looks right
X_train
```

```python
# Building a function to create and compile tf model
def build_model(num_features, learning_rate):

  # Resetting tf for consistent results
  tf.keras.backend.clear_session()
  tf.random.set_seed(0)

  # Building a model using tf Sequential
  model = tf.keras.Sequential()

  # Input layer
  model.add(tf.keras.Input(shape=(num_features,), name='Input'))

  # Output layer
  model.add(tf.keras.layers.Dense(
      units = 3,
      use_bias = True,
      activation = 'softmax',
      kernel_initializer = initializers.RandomNormal(stddev=0.01),
      bias_initializer = initializers.RandomNormal(stddev=0.01),
      name = 'Output'
  ))

  # Getting the optimizer
  optimizer = tf.keras.optimizers.SGD(learning_rate=learning_rate)

  # Compiling the model
  model.compile(
      optimizer = optimizer,
      loss = tf.keras.losses.SparseCategoricalCrossentropy(),
      metrics = ['accuracy']
  )

  return model
```

```python
tf.random.set_seed(1234)
np.random.seed(1234)

# Setting a learning rate, number of epochs, and batch size
learning_rate = 0.0001
num_epochs = 10
batch_size = 32

# Building and compiling the model
model_tf = build_model(
    X_train.shape[1], learning_rate = learning_rate
)

# Training the model using the training dataset and passing the validation
# dataset for validation
model_tf_fit = model_tf.fit(
    x = X_train,
    y = Y_train,
    validation_data = [X_val, Y_val],
    epochs = num_epochs,
    batch_size = batch_size,
    verbose = 1
)
```

```python
# Showing initial model on loss vs epoch plot
plt.rcdefaults()  # In case you changed this earlier
hist = model_tf_fit.history
plt.plot(range(1, num_epochs+1), hist['loss'], label='Training Loss')
plt.plot(range(1, num_epochs+1), hist['val_loss'], label='Validation Loss')
plt.xlabel('Epochs')
plt.ylabel('Loss')
plt.title('Loss vs Epoch')
plt.legend()
plt.show()
```

```python
def build_compile_tf_model(num_features, learning_rate):

    # Clearing and resetting
    tf.keras.backend.clear_session()
    tf.random.set_seed(0)

    # Defining keras Sequential model
    model = tf.keras.Sequential()
    
    # Creating input layer
    model.add(tf.keras.Input(shape=(num_features,),
         name='Input'
    ))

    # Creating output layer
    model.add(tf.keras.layers.Dense(
          activation = None,
          units=1,
          use_bias=True,
          kernel_initializer=tf.ones_initializer,
          bias_initializer=tf.ones_initializer,
          name='Output'
      ))
    
    # SGD optimizer
    optimizer = tf.keras.optimizers.SGD(learning_rate=learning_rate)

    # Compiling the model for mse loss
    model.compile(
        optimizer=optimizer,
        loss='mse', 
    )
    
    return model
```

```python
tf.random.set_seed(1234)
np.random.seed(1234)

# Setting a learning rate, number of epochs, and batch size
learning_rate = 0.0001
num_epochs = 10
batch_size = 32

# Building linear model
model_lin = build_compile_tf_model(
    num_features=X_train.shape[1],
    learning_rate=learning_rate)

# Fitting the model
model_lin_fit = model_lin.fit(
  x = X_train,
  y = Y_train,
  validation_data = [X_val, Y_val],
  epochs=num_epochs,
  batch_size=batch_size,
  verbose=1)
```

```python
# Generating a plot with loss on y-axis and epoch on x-axis
hist = model_lin_fit.history
plt.plot(range(1, 11), hist['loss'], label='Training Loss')
plt.plot(range(1, 11), hist['val_loss'], label='Validation Loss')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.title('Loss vs Epoch')
plt.legend()
plt.show()
```

## Building Time Series Models:

```python
# Removing ticker column and storing in new dataset:
data2 = data.drop('Ticker', axis=1)
# Scaling Data:
scaler = MinMaxScaler(feature_range=(0,1))
scaled_data = scaler.fit_transform(data2)
```

```python
def create_dataset(dataset, look_back=1):
    dataX, dataY = [], []
    for i in range(len(dataset) - look_back):
        dataX.append(dataset[i:(i + look_back)])
        dataY.append(dataset[i + look_back])
    return np.array(dataX), np.array(dataY)

split_idx = int(len(data) * 0.8)
train_data = data.iloc[:split_idx]
test_data = data.iloc[split_idx:]

trainX, trainY = create_dataset(train_data.values, 30)
testX, testY = create_dataset(test_data.values, 30)
```

```python
# Reshape the Data to fit for LSTM:
trainX = np.reshape(trainX, (trainX.shape[0], trainX.shape[1], 7))
testX = np.reshape(testX, (testX.shape[0], testX.shape[1], 7))
```

```python
model = Sequential()
model.add(LSTM(units=50, input_shape=(30, 1)))
model.add(Dense(units=1))
model.compile(loss='mean_squared_error', optimizer='adam')
model.summary()
model.fit(trainX, trainY, epochs=50, batch_size=32)
```

```python
# Predictions:
testPredict = model.predict(testX)
testPredict = scaler.inverse_transform(testPredict)
testY = scaler.inverse_transform(testY.reshape(-1, 1))

# Plot of Performance:
plt.figure(figsize=(12, 6))
plt.plot(testY, label='Actual')
plt.plot(testPredict, label='Predicted')
plt.xlabel('Time')
plt.ylabel('Adj Close Price')
plt.legend()
plt.show()
```
