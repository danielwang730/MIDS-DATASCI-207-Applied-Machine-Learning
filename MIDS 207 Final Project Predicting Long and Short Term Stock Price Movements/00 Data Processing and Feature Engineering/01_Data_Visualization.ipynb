## Imports

```python
import pandas as pd
import numpy as np
import warnings
import matplotlib.pyplot as plt
import seaborn as sns
from mlxtend.plotting import heatmap
```

## Data Retrieval

```python
data = pd.read_csv('data/initial_data.csv')
data.head()
```

```python
# Can also do the following to get the data if running locally
# data = pd.read_csv('SP500_4y_day_data.csv')
```

## Data Visualizations:

```python
# Line Plots for each of our Features:
fig = plt.figure(figsize=(25,25))
i=1
for x in data.columns[2:-1]:
    plt.subplot(2,3,i)
    ax=sns.lineplot(data[x])
    ax.set(xlabel=None)
    plt.title(str(x), loc='center')
    plt.xlabel('Date')
    plt.ylabel('Price ($)')
    i+=1
plt.show()
```

Ignoring Ticker, we find that all charts except Volume demonstrate a similar pattern because after the opening price, the stock market usually has only small variation up and down, so the plots of these prices follow the same pattern. The Volume is a random process for the trade volume in a given day, so it does not follow the same pattern.

```python
# Scatter Plot:
fig = plt.figure(figsize=(10,10))
plt.scatter(data=data,x="Open",y="High")
plt.title("Open vs High")
plt.xlabel("Open")
plt.ylabel("High")
plt.show()
```

The scatterplot above, plotting Open Price against High Price, demonstrates our data has a linear relationship.

```python
# Box Plots:
fig = plt.figure(figsize=(25,25))
i=1
for x in data.columns[2:-1]:
    plt.subplot(2,3,i)
    ax=sns.boxplot(data[x])
    ax.set(xlabel=None)
    plt.title(str(x), loc='center')
    plt.xlabel(None)
    plt.ylabel(None)
    i+=1
plt.show()
```

Our box plots demonstrate that there are many outliers as we are plotting the distributions for all tickers, not just one.
